```python
from langchain_classic.agents import create_react_agent, AgentExecutor
from langchain_classic import hub
```

```python
# Step 2: Pull the ReAct prompt from LangChain Hub
prompt = hub.pull("hwchase17/react")  # pulls the standard ReAct agent prompt
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("""
Answer the following questions as best you can.

You have access to the following tools:

{tools}

Use the following format:

Question: the input question you must answer
Thought: think about what to do
Action: the action to take, must be one of [{tool_names}]
Action Input: the input to the action
Observation: the result of the action
... (this Thought/Action/Action Input/Observation can repeat)
Thought: I now know the final answer
Final Answer: the final answer to the original question

Question: {input}
Thought: {agent_scratchpad}
""")
```

```python
# Step 3: Create the ReAct agent manually with the pulled prompt
agent = create_react_agent(
    llm=llm,
    tools=[tavily_tool],
    prompt=prompt
)
```

```python
# Step 4: Create executor
agent_executor = AgentExecutor(
    agent=agent,
    tools=[tavily_tool],
    handle_parsing_errors=True,
    max_iterations=5,
    verbose=True,
)
```

```python
# Step 5: Invoke the agent
response = agent_executor.invoke({"input": "how to go Pondicherry by bus from kelambakkam"})
print(response)
```

```python
# Access just the output text
response['output']
```

```python
# Step 5: Invoke the agent
response = agent_executor.invoke({"input": "give a detailed preview for last 2026 IPL match play between CSK and MI"})
print(response)
```
